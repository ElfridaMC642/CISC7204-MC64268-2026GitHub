```python
# Step 01
import pandas as pd
```

```python
# Step 02
df = pd.read_csv('forestfires.csv')
```

```python
# Step 03
print(df.head(2))
```

```python
# Step 04
filtered_df = df[df['area'] > 0]
filtered_df.head()
```

```python
# Step 05
print(filtered_df['area'].describe()[['mean', 'min', 'max', 'std']])
```

```python
# Step 06
sorted_df = filtered_df.sort_values(by='area')
sorted_df.tail(20)
```

```python
# Step 07
area_mean = filtered_df['area'].mean()
area_median = filtered_df['area'].median()
print(f"Mean area: {area_mean}")
print(f"Median area: {area_median}")
```

```python
# Step 08
months = df['month'].unique()
months
```

```python
# Step 09
march_count = df[df['month'] == 'mar'].shape[0]
march_count
```

```python
# Step 10
for m in months:
    month_df = df[df['month'] == m]
    fire_count = month_df.shape[0]
    mean_temp = month_df['temp'].mean()
    print(f"Month: {m} | Number of fires: {fire_count} | Mean Temperature: {mean_temp:.2f}°C")
```
