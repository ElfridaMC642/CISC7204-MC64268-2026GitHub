```python
import pandas as pd
```

```python
df=pd.read_csv("London Housing Data.csv")
```

```python
df.head()
```

```python
# Q.5-1) Convert the Datatype of 'Date' column to Date-Time format.
df['date'] = pd.to_datetime(df['date'])
df.head()
```

```python
# Q.5-2) Add a new column ''year'' in the dataframe, which contains years only. (2.B) Add a new column ''month'' as 2nd column in the dataframe, which contains month only.
df['year'] = df['date'].dt.year
df.insert(1, 'month', df['date'].dt.month)
df.head()
```

```python
# Q.5-3) Remove the columns 'year' and 'month' from the dataframe.
df.drop(['year', 'month'], axis=1, inplace=True)
df.head()
```

```python
# Q.5-4) Show all the records where 'No. of Crimes' is 0. And, how many such records are there
df[df['no_of_crimes'] == 0]
```

```python
# Q.5-4) Show all the records where 'No. of Crimes' is 0. And, how many such records are there
len(df[df['no_of_crimes'] == 0])
```

```python
# Q.5-5) What is the maximum 'average_price' per year in england ?
df['year'] = df['date'].dt.year
df_england = df[df['area'] == 'england']

df_england.groupby('year')['average_price'].max()
```

```python
# Q.5-5) What is the minimum 'average_price' per year in england ?
df_england.groupby('year')['average_price'].min()
```

```python
# Q.5-6) What is the Maximum No. of Crimes recorded per area ?
df.groupby('area')['no_of_crimes'].max().sort_values(ascending=False)
```

```python
# Q.5-6) What is the Minimum No. of Crimes recorded per area ?
df.groupby('area')['no_of_crimes'].min().sort_values(ascending=True)
```

```python
# Q.5-7) Show the total count of records of each area, where average price is less than 100000
df[df['average_price'] < 100000]['area'].value_counts()
```
