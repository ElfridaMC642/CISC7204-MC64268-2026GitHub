```python
import pandas as pd
```

```python
df=pd.read_csv("Police Dataset.csv")
```

```python
# Q.3-1) Instruction ( For Data Cleaning ) - Remove the column that only contains missing values.
df.isnull().sum()
```

```python
# Q.3-1) Instruction ( For Data Cleaning ) - Remove the column that only contains missing values.
df.drop(columns='country_name', inplace=True)
df.isnull().sum()
```

```python
# Q.3-2) Question ( Based on Filtering + Value Counts ) - For Speeding , were Men or Women stopped more often ?
speeding_df = df[df['violation'] == 'Speeding']
speeding_df['driver_gender'].value_counts()
```

```python
# Q.3-3) Question ( Groupby ) - Does gender affect who gets searched during a stop ? Question ( mapping + data-type casting )
df.groupby('driver_gender')['search_conducted'].value_counts()
```

```python
# Q.3-4) Question ( mapping + data-type casting ) - What is the mean stop_duration ?
df['stop_duration'].value_counts()
df['stop_duration'] = df['stop_duration'].map({'0-15 Min': 7.5,'16-30 Min': 24,'30+ Min': 45})
df['stop_duration'].mean()
```

```python
# Q.3-5) Question ( Groupby , Describe ) - Compare the age distributions for each violation.
df.groupby('violation')['driver_age'].describe()
```
