```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('movie_scores.csv')
```

```python
df_melted = pd.melt(
    df,
    id_vars=['MovieTitle'],
    value_vars=['AudienceScore', 'Tomatometer'],
    var_name='Type',
    value_name='Scores',)
```

```python
plt.figure(figsize=(10, 6), dpi=200)
sns.set_theme(style='whitegrid')

sns.data = df_melted
ax = sns.barplot(x='MovieTitle', y='Scores', hue='Type', data=df_melted, palette='muted')

plt.title('Movies Scores comparison')
plt.xlabel('Movies')
plt.ylabel('Scores')
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```
