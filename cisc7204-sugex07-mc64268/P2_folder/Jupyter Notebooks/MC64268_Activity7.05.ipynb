```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('anage_data.csv')
```

```python
filtered_df = df[
    (df['Class'] == 'Mammalia') &
    (df['Body mass (g)'] < 200000)]

filtered_df = filtered_df.dropna(
    subset=['Body mass (g)', 'Maximum longevity (yrs)'])
```

```python
sns.set_style('darkgrid')
plt.figure(figsize=(10, 6))

sns.regplot(
    data=filtered_df,
    x='Body mass (g)',
    y='Maximum longevity (yrs)')

plt.xlabel('Body mass (g)')
plt.ylabel('Maximum longevity (yrs)')

plt.show()
```
