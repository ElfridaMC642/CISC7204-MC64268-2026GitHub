```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import squarify
```

```python
mydata = pd.read_csv("water_usage.csv", \
                     index_col=0)
```

```python
labels = mydata['Usage'] \
         + ' (' + mydata['Percentage'].astype('str') + '%)'
```

```python
# Create figure
plt.figure(dpi=200)
# Create tree map
squarify.plot(sizes=mydata['Percentage'], \
              label=labels, \
              color=sns.light_palette('green', mydata.shape[0]))
plt.axis('off')
# Add title
plt.title('Water usage')
# Show plot
plt.show()
```
