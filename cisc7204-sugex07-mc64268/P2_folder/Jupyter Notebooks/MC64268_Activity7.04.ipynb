```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('YouTube.csv')
```

```python
channels = df['Channel'].tolist()
subscribers = df['Subs (in millions)'].tolist()
views = df['Views (in millions)'].tolist()
```

```python
youtube_data = pd.DataFrame({
    'Channel': channels,
    'Subscribers': subscribers,
    'Views': views})
```

```python
youtube_long = youtube_data.melt(
    id_vars='Channel',
    value_vars=['Subscribers', 'Views'],
    var_name='Type',
    value_name='Count')
```

```python
sns.set_style('darkgrid')

g = sns.FacetGrid(
    youtube_long,
    col='Type',
    col_order=['Subscribers', 'Views'],
    sharex=False,
    height=10,
    aspect=0.8)

g.map_dataframe(
    sns.barplot,
    x='Count',
    y='Channel',
    order=channels,
    hue='Type',
    palette={'Subscribers': 'steelblue','Views': 'coral'})

g.set_axis_labels('Count (in millions)', 'YouTube Channels')
g.set_titles('Type = {col_name}')

plt.show()
```
