```python
# With __name__ Convert input before calculating
price_text ="19.95"
quantity_text ="4"
price =float(price_text)
quantity =int(quantity_text)
print(price * quantity)
print(type(price).__name__)
```

```python
# Without __name__ Convert input before calculating
price_text ="19.95"
quantity_text ="4"
price =float(price_text)
quantity =int(quantity_text)
print(price * quantity)
print(type(price))
```

```python
#Comprehensions build lists concisely
prices = [1200, 30, 80, 350]
discounted = [price * 0.9 for price in prices]
expensive = [price for price in prices if price >= 300]
print(discounted)
print(expensive)
```

```python
order = {"product": "Laptop", "quantity": 3}
print(order.get("product")) # with get if no product in order will show NONE as result, otherwise will be ERROR
print(order.get("discount", 0))
order["quantity"] += 1
order["region"] = "North"
print(order)
```

```python
#Collections can be nested
customer = {
 "name": "Ada",
 "orders": [
 {"id": 101, "total": 3600},
 {"id": 102, "total": 300}
 ]
}
print(customer["orders"][1]["total"])
print(sum(o["total"] for o in customer["orders"]))
```
