```python
# Step 02
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Step 03
movie_scores = pd.read_csv("movie_scores.csv")
```

```python
# Step 04 & 05 & 06
plt.figure(figsize=(10, 6), dpi=150)

x = np.arange(len(movie_scores["MovieTitle"]))
width = 0.35  # Width of the bars

plt.bar(x - width / 2, movie_scores["Tomatometer"], width, label="Tomatometer")
plt.bar(x + width / 2, movie_scores["AudienceScore"], width, label="Audience Score")

ax = plt.gca()

ax.set_xticks(x)
ax.set_xticklabels(movie_scores["MovieTitle"], rotation=15, ha="right")
ax.set_yticks(np.arange(0, 101, 20))
ax.set_yticklabels(["0%", "20%", "40%", "60%", "80%", "100%"])
ax.set_yticks(np.arange(0, 101, 5), minor=True)

ax.yaxis.grid(which="major")
ax.yaxis.grid(which="minor", linestyle="--")

plt.title("Movie comparison")
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()
```
