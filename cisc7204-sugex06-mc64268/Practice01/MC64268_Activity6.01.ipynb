```python
# Step 02
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Step 03
googl_df = pd.read_csv("GOOGL_data.csv")
fb_df = pd.read_csv("FB_data.csv")
aapl_df = pd.read_csv("AAPL_data.csv")
amzn_df = pd.read_csv("AMZN_data.csv")
msft_df = pd.read_csv("MSFT_data.csv")
```

```python
googl_df["date"] = pd.to_datetime(googl_df["date"])
fb_df["date"] = pd.to_datetime(fb_df["date"])
aapl_df["date"] = pd.to_datetime(aapl_df["date"])
amzn_df["date"] = pd.to_datetime(amzn_df["date"])
msft_df["date"] = pd.to_datetime(msft_df["date"])
```

```python
# Step 04
plt.figure(figsize=(10, 6), dpi=150)
plt.plot(googl_df["date"], googl_df["close"], label="Google")
plt.plot(fb_df["date"], fb_df["close"], label="Facebook")
plt.plot(aapl_df["date"], aapl_df["close"], label="Apple")
plt.plot(amzn_df["date"], amzn_df["close"], label="Amazon")
plt.plot(msft_df["date"], msft_df["close"], label="Microsoft")


plt.title("Stock trend")
plt.xlabel("Date")
plt.ylabel("Stock price ($)")
plt.legend(loc="upper left")
plt.grid(True)

plt.show()
```
