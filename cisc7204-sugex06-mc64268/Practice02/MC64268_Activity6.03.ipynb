```python
# Step 01
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
bills = sns.load_dataset('tips')
```

```python
# Step 02
bills_matrix = bills.pivot_table(values="total_bill", index="day", columns="smoker", aggfunc="sum")
```

```python
# Step 03 & 04
plt.figure(figsize=(8, 6), dpi=100)

plt.bar(bills_matrix.index, bills_matrix["Yes"], label="Smoker")
plt.bar(
    bills_matrix.index,
    bills_matrix["No"],
    bottom=bills_matrix["Yes"],
    label="Non-Smoker",)

plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")
plt.title("Restaurant Daily Sales by Smoker/Non-Smoker")
plt.legend(title="Smoker")
plt.show()
```
