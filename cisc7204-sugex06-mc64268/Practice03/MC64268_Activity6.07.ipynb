```python
# Step 01
import os
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
%matplotlib inline
```

```python
# Step 03 & 04
img1 = mpimg.imread("photo-1.jpg")
img2 = mpimg.imread("photo-2.jpg")
img3 = mpimg.imread("photo-3.jpg")
img4 = mpimg.imread("photo-4.jpg")

fig, ax = plt.subplots(2, 2, figsize=(8, 8))

ax[0, 0].imshow(img1)
ax[0, 0].set_title("coast")
ax[0, 0].axis("off")


ax[0, 1].imshow(img2)
ax[0, 1].set_title("beach")
ax[0, 1].axis("off")

ax[1, 0].imshow(img3)
ax[1, 0].set_title("building")
ax[1, 0].axis("off")

ax[1, 1].imshow(img4)
ax[1, 1].set_title("city at night")
ax[1, 1].axis("off")

plt.tight_layout()
plt.show()
```
