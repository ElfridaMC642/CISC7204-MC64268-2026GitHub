```python
# Step 01
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Step 02
data = pd.read_csv('anage_data.csv')

data = data[np.isfinite(data['Body mass (g)']) & np.isfinite(data['Maximum longevity (yrs)'])]
aves = data[(data['Class'] == 'Aves') & (data['Body mass (g)'] < 20000)]

mass = aves['Body mass (g)']
longevity = aves['Maximum longevity (yrs)']
```

```python
# Step 03
fig = plt.figure(figsize=(8, 8), layout='constrained')
fig.suptitle('Scatter plot with marginal histograms', fontsize=14)

gs = fig.add_gridspec(4, 4)
ax_histx = fig.add_subplot(gs[0, :3])
ax_scatter = fig.add_subplot(gs[1:, :3])
ax_histy = fig.add_subplot(gs[1:, 3])

ax_histx.hist(mass, bins=20, edgecolor='black', linewidth=0.5)
ax_histx.set_xticks([]) 

ax_scatter.scatter(mass, longevity, alpha=0.7)
ax_scatter.set_xlabel('Body mass in grams')
ax_scatter.set_ylabel('Maximum longevity in years')

ax_histy.hist(longevity, bins=20, orientation='horizontal', edgecolor='black', linewidth=0.5)
ax_histy.set_yticks([])

plt.show()
```
